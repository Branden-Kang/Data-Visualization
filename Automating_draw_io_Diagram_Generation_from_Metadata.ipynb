```
YAML Mapping → Metadata Tables in DuckDB → Python Script → draw.io XML → .drawio File
```

````python
import duckdb
import xml.etree.ElementTree as ET

def fetch_metadata():
    con = duckdb.connect('metadata.duckdb')
    return con.execute("SELECT * FROM mappings").fetchall()

def create_drawio_xml(metadata):
    diagram = ET.Element('diagram')
    for row in metadata:
        shape = ET.SubElement(diagram, 'mxCell', {
        'id': row["id"],
        'value': row["label"],
        'style': 'shape=rectangle;fillColor=#dae8fc;',
        'vertex': '1',
        'parent': '1'
        })
        geometry = ET.SubElement(shape, 'mxGeometry', {
        'x': str(row['x']),
        'y': str(row['y']),
        'width': '120',
        'height': '40',
        'as': 'geometry'
        })
    return ET.tostring(diagram, encoding='unicode')

metadata = fetch_metadata()
diagram_xml = create_drawio_xml(metadata)

with open('output.drawio', 'w') as f:
    f.write(diagram_xml)
````
