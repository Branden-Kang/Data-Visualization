# Step 2: Your First Streamlit App

````python
# app.py
import streamlit as st
import pandas as pd
import numpy as np

st.title("Hello, Streamlit! 👋")
st.write("This is your first Streamlit app 🚀")

# Generate some random data
data = pd.DataFrame(
    np.random.randn(10, 2),
    columns=["X", "Y"]
)

st.line_chart(data)
````

```
streamlit run app.py
```
